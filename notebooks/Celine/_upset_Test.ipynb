```python
metabolites = ('Adelie', 'Chinstrap', 'Gentoo')
weeks = {
    'WK5': np.array([73, 34, 61]),
    'WK12': np.array([73, 34, 58]),
}
width = 0.6  # the width of the bars: can also be len(x) sequence


fig, ax = plt.subplots()
bottom = np.zeros(3)

for sex, sex_count in weeks.items():
    p = ax.bar(metabolites, sex_count, width, label=sex, bottom=bottom)
    bottom += sex_count

    ax.bar_label(p, label_type='center')

ax.set_title('metabolites per weeks')
ax.legend()

plt.show()
```
